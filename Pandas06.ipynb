```python
import pandas as pd
algeria_wather={
    "city" : ["biskra","batna","blida","annaba"] , 
    "temperature" : [35,28,24,26] , 
    "humidity" : [7,15,20,60]
    }
df01=pd.DataFrame(algeria_wather)
df01
```

```python
usa_wather={
    "city" : ["new york" ,"chicago" ,"california","washington"],
    "temperature" : [28,35,26,24],
    "humidity" : [20,60,15,7],
}
df02=pd.DataFrame(usa_wather)
df02
```

```python
df=pd.concat([df01,df02])
# df=pd.concat([df01,df02] , ignore_index=True)   to ignore the index and it count 0,1,2,3,4,5,6,...
df
```

```python
df=pd.concat([df01,df02] , keys=["Algeria" , "USA"])
df
```

```python
temperature_data=pd.DataFrame({
    "city" : ["new york" ,"chicago" ,"california","washington"],
    "temperature" : [28,35,26,24],
})
temperature_data
```

```python
windspeed_data=pd.DataFrame({
    "city" : ["chicago" ,"new york" ,"california","washington"],
    "windspeed" : [28,35,26,24],
} ,
index=[1,0,2,3])
windspeed_data
```

```python
df=pd.concat([temperature_data , windspeed_data["windspeed"]] , axis="columns")  #  axis="rows" : axis=0
df
```

```python
df=pd.concat([temperature_data,windspeed_data["windspeed"]] , axis="rows")  #  axis="rows" : axis=0
df
```

```python
event=pd.Series(["humdii","dry","rain","sunny"] , name="event")
event
```

```python
df=pd.concat([windspeed_data,event] , axis=1)
df
```
