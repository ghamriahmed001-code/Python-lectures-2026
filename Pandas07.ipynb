```python
import pandas as pd
temperature_data=pd.DataFrame({
    "city" : ["biskra","batna","alger","adrar"] ,
    "temperature" :[35,28,24,26] 
})
temperature_data
```

```python
windspeed_data=pd.DataFrame({
    "city" : ["biskra","batna","blida","annaba"] ,
    "windspeed" : [28,35,26,24]
})
windspeed_data
```

```python
total_data=pd.merge(temperature_data , windspeed_data , on="city")  #   by defualt : how="inner"
total_data
#علاش بدا بها الترتيب بالذات ؟؟؟؟؟؟  :
```

```python
total=pd.merge(temperature_data , windspeed_data , on="city" , how="outer")
total
```
