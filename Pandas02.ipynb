```python
import pandas as pd
excel_data01 = {
    "ticker"  : ["GOOGL","WMT","MSFT","RIL","TATA"],
    "eps"     : [27.82,4.61,-1,"not available",5.6] ,
    "revenue" : [87,484,85,50,-1] ,
    "price"   : [845,65,64,1023,"n.a."] ,
    "people"  : ["larry page","n.a.","bill gates","mukesh ambani","ratan atat"]
}
df01=pd.DataFrame(excel_data01)
df01.replace(
    {
        "eps" : ["not available"] ,
        "revenue" : [-1] ,
        "price" : ["n.a."],
        "people" : ["n.a."]
    },
    None ,
    inplace=True ,
)
excel_data02={
    "name"    : ["Ahmed" ,"Osama" ,"Omar" ,"Sami"] ,
    "country" : ["Algeria" ,"Egypt" ,"Saudi" ,"Jordan"],
    "score"   : [77,82,95,67]
}
df02=pd.DataFrame(excel_data02)
with pd.ExcelWriter("excel01_&_02.xlsx") as writer :
    df01.to_excel(writer,sheet_name="data01" , index=False)
    df02.to_excel(writer,sheet_name="data02" , index=False)
```
